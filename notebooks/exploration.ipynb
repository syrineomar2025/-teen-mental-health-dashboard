Permet de visualiser un aperçu des données.

### Statistiques descriptives
Résumé des variables numériques.

```python
df.tail()
df.shape
df.dtypes
df.info()
df.describe()
```

### Vérification des doublons
Nombre de lignes dupliquées.

```python
df.duplicated().sum()
```

### Suppression des doublons

```python
df=df.drop_duplicates()
```

### Valeurs manquantes
Nombre de valeurs nulles par colonne.

```python
df.isnull().sum()
```

### Suppression des valeurs manquantes

```python
df.dropna()
```

### Remplacement des valeurs manquantes
- Moyenne pour les variables numériques
- Mode pour les variables catégorielles

```python
df.fillna(df.mean(numeric_only=True))
```

```python
df.fillna(df.mode().iloc[0])
```

### Filtrage simple
Personnes avec âge > 16

```python
df[df["age"] > 16]
```

### Filtrage avec condition multiple

```python
df[(df["age"] > 16) & (df["daily_social_media_hours"] > 4)]
```

### Filtrage par catégories

```python
df[df["gender"].isin(["Male", "Female"])]
```

### Tri par âge

```python
df.sort_values(by="age")
```

```python
df.sort_values(by="daily_social_media_hours", ascending=False)
```

### Distribution du genre

```python
df["gender"].value_counts()
```

### Distribution en pourcentage

```python
df["gender"].value_counts(normalize=True)
```

### Moyenne du temps sur réseaux sociaux par genre

```python
df.groupby("gender")["daily_social_media_hours"].mean()
```

### Moyenne du sommeil par niveau de stress

```python
df.groupby("stress_level")["sleep_hours"].mean()
```

### Moyenne du sommeil par niveau de stress

```python
df.groupby(["gender", "stress_level"])["daily_social_media_hours"].mean()
```
